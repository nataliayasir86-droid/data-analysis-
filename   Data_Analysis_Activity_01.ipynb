```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df1 = pd.read_csv("pakistani_ecommerce_dataset.csv")

df1.head()
```

```python
df1.info()
```

```python
df1.columns
```

```python
#Q1. Which is the most popular product category?

df1["product_category"].value_counts()
```

```python
#Q2 What is the average order value?

df1["order_value"].mean()
```

```python
#Q3. Which city has the highest number of orders?
df1["city"].value_counts()
```

```python
#Q4. What is the most used payment method?
df1["payment_method"].value_counts()
```

```python
#Q5. Draw a bar chart showing average order value by city.
city_average = df1.groupby("city")["order_value"].mean()

city_average.plot(kind="bar")

plt.title("Average Order Value by City")
plt.xlabel("City")
plt.ylabel("Average Order Value")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df2 = pd.read_csv("student_lifestyle_dataset.csv")

df2.head()
```

```python
df2.info()

df2.columns
```

```python
#What is the average number of study hours?
df2["study_hours"].mean()
```

```python
#Q2: Who sleeps more on average – males or females?
df2.groupby("gender")["sleep_hours"].mean()
```

```python
#Q3  Is there a relationship between screen time and sleep hours?
df2["screen_time"].corr(df2["sleep_hours"])

```

```python
#Q4: Draw a histogram of daily transport cost.
plt.hist(df2["daily_transport_cost"], bins=10)

plt.title("Daily Transport Cost Distribution")
plt.xlabel("Daily Transport Cost")
plt.ylabel("Number of Students")

plt.show()
```
